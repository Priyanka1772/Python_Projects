## Diwali Sales Analysis

```python
## Import relevant libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## Load the dataset

df = pd.read_csv('Diwali Sales Data.csv')
df
```

```python
## Data Inspection 

df.shape
```

```python
## Data View

df.head()
```

```python
## Data Inspection

df.info()
```

```python
## Data Inspection

df.describe()
```

```python
## Drop unrelated/blank columns

df.drop(['Status', 'unnamed1'], axis = 1, inplace = True)
```

```python
df.info()
df.shape
```

```python
## Check null values

pd.isnull(df).sum()
```

```python
## Drop null values

df.dropna(inplace = True)
pd.isnull(df).sum()
df.shape
```

```python
## Chnage data type of a column 

df['Amount'] = df['Amount'].astype('int')
df['Amount'].dtypes
```

```python
## View all column names

df.columns
```

```python
# Changes wont be saved oroginally cz 'inplace' is not used

df.rename(columns = {'Marital_Status': 'Married_Or_Not'})
```

```python
df.columns
```

## Exploratory Data Analysis
### Gender

```python
# A bar chart for Gender wrt to its count

graph_1 = sns.countplot(x = 'Gender', data = df)

# To activate data labels
for bars in graph_1.containers:
    graph_1.bar_label(bars)
```

```python
## A bar chart for Gender VS Total amount

graph_2_data = df.groupby(['Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)
```

```python
sns.barplot(x = 'Gender', y = 'Amount', data = graph_2_data)
```

#### Insights: From above graphs we can see that most of the buyers are females and even the purchasing power of females are greater than men

### Age

```python
graph_3 = sns.countplot(x = 'Age Group', data = df, hue = 'Gender')

for bars in graph_3.containers:
    graph_3.bar_label(bars)
```

```python
Top_n = int(input('Enter Top \'n\' value (3, 5, 7, 10):'))

graph_4_data = df.groupby(['State'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(Top_n)
```

```python
sns.set(rc = {'figure.figsize': (18,5)})
```

```python
## Total Amount vs Age Group

graph_4 = sns.barplot(x = 'State', y = 'Orders', data = graph_4_data)

for bars in graph_4.containers:
    graph_4.bar_label(bars)
```

### State

```python
# Total number of orders from Top 10 States

Top_n = int(input('Enter Top \'n\' value (3, 5, 7, 10):'))

graph_5_data = df.groupby(['State'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(Top_n)

sns.set(rc = {'figure.figsize': (20,5)})
graph_5 = sns.barplot(x = 'State', y = 'Amount', data = graph_5_data)
```

```python
# Total amount/Sales from Top 10 States

graph_6 = sns.countplot(x = 'Marital_Status', data = df)

sns.set(rc = {'figure.figsize': (5,5)})
for bars in graph_6.containers:
    graph_6.bar_label(bars)
```

#### Insights: From above graphs we can see that most of the orders & total sales/amount are from Uttar Pradesh, Maharashtra and Karnataka respectively

### Marital Status

```python
graph_7_data = df.groupby(['Marital_Status', 'Gender'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False)

graph_7 = sns.barplot(x = 'Marital_Status', y = 'Amount', data = graph_7_data, hue = 'Gender')
```

#### Insights: From above graphs we can see that most of the buyers are married (women) and they have high purchasing power

### Occupation

```python
sns.set(rc = {'figure.figsize':(20,5)})
graph_8 = sns.countplot(x = 'Occupation', data = df)

for bars in graph_8.containers:
    graph_8.bar_label(bars)
```

```python
Top_n = int(input('Enter a Top \'n\' value (3, 5, 7, 10, 15):'))

graph_9_data = df.groupby(['Occupation'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending =  False).head(Top_n)

sns.set(rc = {'figure.figsize':(20,5)})
graph_9 = sns.barplot(x = 'Occupation', y = 'Amount', data = graph_9_data)
```

#### Insights: From above graphs we can see that most of the buyers are working in IT, Healthcare and Aviation sector

### Product Category 

```python
sns.set(rc = {'figure.figsize':(25,5)})
graph_10 = sns.countplot(x = 'Product_Category', data = df)

for bars in graph_10.containers:
    graph_10.bar_label(bars)
```

```python
Top_n = int(input('Enter a Top \'n\' value (3, 5, 7, 10, 15, 18): '))

graph_11_data = df.groupby(['Product_Category'], as_index = False)['Amount'].sum().sort_values(by = 'Amount', ascending = False).head(Top_n)

sns.set(rc = {'figure.figsize': (25,8)})

graph_11 = sns.barplot(x = 'Product_Category', y = 'Amount', data = graph_11_data)
```

#### Insights: From above graphs we can see that most of the sold products are from Food, Clothing and Electronics category

```python
# Top 10 most sold Products

Top_n = int(input('Enter a Top \'n\' value (3, 5, 7, 10): '))

graph_12_data = df.groupby(['Product_ID'], as_index = False)['Orders'].sum().sort_values(by = 'Orders', ascending = False).head(Top_n)

sns.set(rc = {'figure.figsize': (20,5)})

graph_12 = sns.barplot(x = 'Product_ID', y = 'Orders', data = graph_12_data)

for bars in graph_12.containers:
    graph_12.bar_label(bars)
```

## Conclusion: 
Married women age group 26-35 yrs from UP, Maharastra and Karnataka working in IT, Healthcare and Aviation are more likely to buy products from Food, Clothing and Electronics category
